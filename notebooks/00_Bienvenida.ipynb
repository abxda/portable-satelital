````python
# 2) Prueba geoespacial real: crear, guardar, releer y dibujar un ráster
import numpy as np, rasterio
from rasterio.transform import from_origin
import matplotlib.pyplot as plt

# Un "NDVI" sintético de 200x200 píxeles en CRS de México (EPSG:6372)
rng = np.random.default_rng(42)
x, y = np.meshgrid(np.linspace(-3, 3, 200), np.linspace(-3, 3, 200))
ndvi = (np.sin(x) * np.cos(y) + rng.normal(0, 0.1, x.shape)).astype("float32")

with rasterio.open("ndvi_demo.tif", "w", driver="GTiff", height=200, width=200,
                   count=1, dtype="float32", crs="EPSG:6372",
                   transform=from_origin(2500000, 1200000, 30, 30)) as dst:
    dst.write(ndvi, 1)

with rasterio.open("ndvi_demo.tif") as src:
    data = src.read(1)
    print(f"GeoTIFF leído: {src.width}x{src.height}, CRS {src.crs}")

plt.figure(figsize=(6, 5))
plt.imshow(data, cmap="RdYlGn")
plt.colorbar(label="NDVI sintético")
plt.title("Si ves este mapa, GDAL y PROJ funcionan ✅")
plt.show()
````
